```python
# Datsun PL510
x=0.97
y=8.7
w=5
lr=0.1

pred=w*x
print("pred=",pred)

error=(pred-y)**2
print("error=",error)

stepness=2*(pred-y)*x
print("stepness=",stepness)

print("-"*30)

w_new=w-lr*stepness
pred_new=w_new*x
print("pred_new=",pred_new)

error_new=(pred_new-y)**2
print("error_new=",error_new)
```
